```python
import numpy as np
import pandas as pd
```

```python
#PART 1 NUMPY 1D ARRAY
temperatures_w1=np.array([22.5,25.3,20.8,23.4,26.1,24.8,21.9])
print("WEEK 1 TEMPERATURES")
print(temperatures_w1)
```

```python
print("Shape:", temperatures_w1.shape)
print("Data Type:", temperatures_w1.dtype)
print("Number of Elements:", temperatures_w1.size)
```

```python
#celcius to fahrenheit
temperatures_fahrenheit = (temperatures_w1 * 9/5) + 32

print("Temperatures in Fahrenheit:")
print(temperatures_fahrenheit)
```

```python
#maximum
maximum_temperature = np.max(temperatures_w1)

print("Maximum Temperature:", maximum_temperature)
```

```python
#minimuum
minimum_temperature = np.min(temperatures_w1)

print("Minimum Temperature:", minimum_temperature)
```

```python
#mean
mean_temperature = np.mean(temperatures_w1)

print("Mean Temperature:", mean_temperature)
print("Mean Temperature:", round(mean_temperature, 2))
```

```python
#slicing
first_three_days = temperatures_w1[:3]

print("First three days:")
print(first_three_days)

weekend = temperatures_w1[-2:]

print("Weekend temperatures:")
print(weekend)

middle_three_days = temperatures_w1[2:5]

print("Middle three days:")
print(middle_three_days)
```

```python
#NUMPY 2D ARRAY
temperatures = np.array([
    [22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9],
    [19.2, 22.5, 21.3, 24.0, 23.5, 22.8, 20.1]
])

print("Temperature Data:")
print(temperatures)
```

```python
print("Shape:", temperatures.shape)
print("Data Type:", temperatures.dtype)
print("Total Elements:", temperatures.size)
```

```python
week1 = temperatures[0]

print("Week 1:")
print(week1)

week2 = temperatures[1]

print("Week 2:")
print(week2)

weekends = temperatures[:, -2:]

print("Weekend temperatures:")
print(weekends)
```

```python
#PANDAS DATAFRAME
transactions = pd.DataFrame({
    "TransactionID": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],

    "ProductCategory": [
        "Electronics", "Clothing", "Electronics", "Furniture",
        "Clothing", "Electronics", "Furniture", "Clothing",
        "Furniture", "Electronics"
    ],

    "Region": [
        "North", "South", "North", "East", "West",
        "North", "East", "West", "South", "North"
    ],

    "Amount": [
        200, 150, 300, 450, 200,
        250, 300, 180, 350, 400
    ]
})

print(transactions)

```

```python
#Data Exploration
print(transactions.head())
```

```python
print(transactions.tail())
```

```python
print("Shape:", transactions.shape)
```

```python
print("Column Names:")
print(transactions.columns)
```

```python
print("Data Types:")
print(transactions.dtypes)
```

```python
#select columns
print(transactions[["ProductCategory", "Amount"]])
```

```python
#Last Three Columns
print(transactions.iloc[:, -3:])
```

```python
# Filter North and Amount > 200

filtered_data = transactions[
    (transactions["Region"] == "North") &
    (transactions["Amount"] > 200)
]

print(filtered_data)
```

```python
# Value counts

print(transactions["ProductCategory"].value_counts())
```

```python
# Unique regions
print(transactions["Region"].unique())
```

```python
# Mean amount by region
region_mean = transactions.groupby("Region")["Amount"].mean()

print(region_mean)
```

```python
# PART 4: MANIPULATING DATAFRAME
# Change TransactionID 102 amount to 165

transactions.loc[
    transactions["TransactionID"] == 102,
    "Amount"
] = 165

print(transactions)
```

```python
# Add Discount column

transactions["Discount"] = transactions["Amount"] * 0.10

print(transactions)
```

```python
# Remove TransactionID 109

transactions = transactions[
    transactions["TransactionID"] != 109
]

print(transactions)
```

```python
# PART 5: PANDAS SERIES

marks = pd.Series(
    [95, 92, 89, 85, 80],
    index=["Rank1", "Rank2", "Rank3", "Rank4", "Rank5"]
)

print(marks)
```

```python
# Indexing and slicing

print(marks.iloc[0])

print(marks.loc[["Rank1", "Rank2", "Rank3"]])

print(marks.iloc[2])

print(marks[marks > 90])
```

```python
# Manipulating Series

marks.loc["Rank1"] = 100

print(marks)
```

```python
# Remove Rank5

marks = marks.drop("Rank5")

print(marks)
```

```python
# Calculate CGPA

cgpa = marks / 10

print(cgpa)
```
